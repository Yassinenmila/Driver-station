# **Statistique Descriptif de la data frame** 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv('../data/car-price.csv')

print(df.head())

print(df.info)

print(df.shape[0])

print(df.shape[1])

print(df.describe())

print(df.isna().sum())

print(df.duplicated().sum())

print(df.dtypes)

```

# **Explotation graphiques**

## 1. distribution des prix 


La distribution montre la répartition des prix des véhicules dans le dataset. On observe que la majorité des véhicules se concentre dans une certaine plage de prix, tandis que quelques véhicules présentent des prix beaucoup plus élevés. Ces valeurs seront vérifiées lors de l'analyse des valeurs aberrantes.

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="selling_price",
    bins=30,
    kde=True
)

plt.title("Distribution du prix des voitures")
plt.xlabel("Prix de vente")
plt.ylabel("Nombre de voitures")

plt.show()
```

## 2. Distribution de kilometrage


La distribution du kilométrage montre une concentration des véhicules sur certaines valeurs, avec quelques observations présentant un kilométrage élevé. Ces valeurs seront examinées afin de déterminer si elles correspondent à des situations réelles ou à des valeurs aberrantes.

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="km_driven",
    bins=30,
    kde=True
)

plt.title("Distribution du kilométrage")
plt.xlabel("Kilométrage")
plt.ylabel("Nombre de voitures")

plt.show()
```

## 3. Annee & Prix

Le graphique permet d'observer la relation entre l'année du véhicule et son prix. Les véhicules plus récents semblent généralement présenter des prix différents des véhicules plus anciens. Cette variable sera donc conservée pour la modélisation.

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="year",
    y="selling_price"
)

plt.title("Relation entre l'année et le prix")
plt.xlabel("Année du véhicule")
plt.ylabel("Prix de vente")

plt.show()
```

## 4. Relation entre Kilometrage et prix 

Le graphique permet d'étudier la relation entre le kilométrage et le prix de vente. On observe la dispersion des prix selon le kilométrage, ce qui permet d'évaluer l'intérêt de cette variable pour la prédiction.

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="km_driven",
    y="selling_price"
)

plt.title("Relation entre le kilométrage et le prix")
plt.xlabel("Kilométrage")
plt.ylabel("Prix de vente")

plt.show()
```

## 5. Matrice de correlation

Cette matrice permet d’analyser les relations entre les différentes variables numériques du dataset. Les coefficients de corrélation permettent notamment d’observer quelles variables présentent une relation avec le prix de vente.

```python
correlation = df.corr(numeric_only=True)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Matrice de corrélation")

plt.show()
```

## 6. Boxplot du prix

Ce boxplot permet d’identifier les valeurs potentiellement aberrantes du prix de vente. Les observations situées en dehors des limites du graphique seront ensuite analysées afin de déterminer si elles correspondent à des valeurs réellement inhabituelles ou à des erreurs dans les données.

```python
plt.figure(figsize=(10, 4))

sns.boxplot(
    data=df,
    x="selling_price"
)

plt.title("Détection des valeurs aberrantes du prix")
plt.xlabel("Prix de vente")

plt.show()
```

## 7. boxplot de kilometrage

Ce boxplot permet d’identifier les valeurs potentiellement aberrantes du kilométrage. Ces observations seront vérifiées avant toute décision de suppression.

```python
plt.figure(figsize=(10, 4))

sns.boxplot(
    data=df,
    x="km_driven"
)

plt.title("Détection des valeurs aberrantes du kilométrage")
plt.xlabel("Kilométrage")

plt.show()
```

## 8. boxplot pour Year



```python
plt.figure(figsize=(10, 4))

sns.boxplot(
    data=df,
    x="year"
)

plt.title("Distribution de l'année des véhicules")
plt.xlabel("Année")

plt.show()
```
